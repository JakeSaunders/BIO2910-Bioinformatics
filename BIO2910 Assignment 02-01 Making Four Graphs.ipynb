### Selecting one columnBefore you can graph a column you have to select it. There are two ways.**Way 1 — square brackets,** using positions: `variable[ rows , columns ]`

```r
dat[ , 1 ]      # every row, column 1
```

**Way 2 — the dollar sign,** using the column's name: `variable$column.name`This is the one you will use most often.

```r
dat$Sepal.Length
```

### Make a histogram of sepal length

```r
hist(  # x is the argument for the values to be plotted  x = dat$Sepal.Length,  # roughly how many bars to divide the data into  breaks = 10,  # bar color and border color  col = "gold", border = "navyblue",  # main title  main = "Sepal Length of Three Species of Iris",  # x-axis label  xlab = "Sepal Length (cm)")
```

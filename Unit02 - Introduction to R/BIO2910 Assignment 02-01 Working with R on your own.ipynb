### Vectors (lists)

R is really good at storing and processing lists of things. To make a list — R calls it a
**vector** — of consecutive numbers, use the first and last number separated by a `:`

````r
1:100
````

You can also make a vector of non-consecutive numbers by putting them inside the combine (or
concatenate) function `c(...)`. The three dots are shorthand for "a list of things separated by
commas."

````r
c(1, 7, 9, 100, 80, 280, 2, 59)
````

`c()` is one of the simplest functions in R, and like every function you can get help on it by
putting a `?` in front of it.

Read through the documentation and see if you have any questions. The information is very
detailed, but every R help page has the same format — so the secret to figuring out how to do
new things in R is often just reading the documentation.

````r
?c
````

Just as you assigned a single number to a variable, you can assign a whole list to one.

````r
my.numbers <- c(1, 7, 9, 100, 80, 280, 2, 59)

my.numbers
````

````r
my.strings <- c("You", "Even", "A", "of", "These", "My", "combined")

my.other.strings <- c("Can", "Make", "List", "strings.", "Are", "Strings", "together.")

my.strings
````

````r
my.other.strings
````

If a list is made up entirely of numbers you can do math with the whole list at once. The code

You can even do math with two numeric lists. The code below divides every number in `my.numbers`

You can also assign the output of code to a new variable.

Make sure the value of the thing on the right is what you want the value of the thing on the
left to become. The assignment operator (`<-`) is an arrow for exactly this reason. The equals
sign also works this way, but it is used for other things in R too, so it is best not to use the
equals sign for both jobs.

````r
my.new.numbers <- my.numbers / 10:17

my.new.numbers
````

### Functions that work on text

### Subsetting a data frame

There are two ways to select parts of a data frame. The first selects rows and columns by
**number**:

```
DNase[ numeric-list-of-rows , numeric-list-of-columns ]
```

The code below selects rows 20 through 30.

````r
DNase[20:30, ]
````

You can also subset by columns. This selects columns 2 and 3.

````r
DNase[ , 2:3]
````

Or by both at once.

````r
DNase[40:45 , 2:3]
````

Before you move on, notice the **commas** in the lines above. Comma placement is very important
in R, as you may already have discovered with `paste()`. A good first step when troubleshooting
is to double-check your commas.

Spaces and line returns, on the other hand, generally don't matter. Run all three cells below and
notice that each gives the same result.

````r
DNase[40:45, 2:3]
````

````r
      DNase[              40:45      ,      2:3       ]
````

````r
DNase[
    40:45,
    2:3
]
````

## Part 2) Histogram

Look up the help for `hist` first.

```r
?hist
```

### Selecting one column

Before you can graph a column you have to select it. There are two ways.

**Way 1 — square brackets,** using positions: `variable[ rows , columns ]`

```r
dat[ , 1 ]      # every row, column 1
```

```r
dat[ , 1:2 ]    # every row, columns 1 and 2
```

```r
dat[ 1:10 , 1:2 ]   # rows 1-10, columns 1 and 2
```

**Way 2 — the dollar sign,** using the column's name: `variable$column.name`

This is the one you will use most often.

```r
dat$Sepal.Length
```

### Make a histogram of sepal length

```r
hist(
  # x is the argument for the values to be plotted
  x = dat$Sepal.Length,
  # roughly how many bars to divide the data into
  breaks = 10,
  # bar color and border color
  col = "gold", border = "navyblue",
  # main title
  main = "Sepal Length of Three Species of Iris",
  # x-axis label
  xlab = "Sepal Length (cm)"
)
```

## Part 3) Barplot

`colMeans()` calculates the mean of each column of a data frame. Columns 1 through 4 of `dat`
are the four measurements; column 5 is Species, which has no mean.

```r
colMeans(dat[1:4])
```

```r
?barplot
```

```r
barplot(
  # make bar heights equal to the column means
  height = colMeans(dat[1:4]),
  # use the c() function to provide labels for the bars
  names.arg = c("Sepal Length", "Sepal Width", "Petal Length", "Petal Width"),
  # use the c() function to pick colors for the bars
  col = c("blue", "blue", "red", "red"),
  main = "This is my main title",
  xlab = "This is my x-axis label",
  ylab = "This is my y-axis label"
)
```

## Part 4) Boxplot

The `~` (tilde) here means "by" — read `dat$Sepal.Length ~ dat$Species` as

## Saving your graphs

In RStudio you saved plots with **Export → Save as Image**. Colab has no Plots panel, so there
are two ways instead.

**The quick way:** right-click any graph in the output and choose **Save image as…**

**The reproducible way:** wrap the plotting code in `png()` and `dev.off()`. The file lands in
the Files sidebar (the 📁 folder icon on the left), where you can download it.

```r
# Example: save the scatter plot as a .png file
png(filename = "my_scatterplot.png", width = 800, height = 600)

plot(
  x = dat$Petal.Width, y = dat$Petal.Length,
  col = dat$Species, pch = 19,
  main = "Petal Width and Length are Proportional",
  xlab = "Petal Width (cm)", ylab = "Petal Length (cm)", cex = 2
)

dev.off()   # this closes the file -- the plot is NOT saved until you run this

# confirm the file was created
list.files(pattern = "png")
```
